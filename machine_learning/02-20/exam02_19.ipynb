```python
import tensorflow as tf
input_layer = tf.keras.layers.Input(shape=(2,))
dense1 = tf.keras.layers.Dense(units=2,
                               activation='sigmoid')
dense2 = tf.keras.layers.Dense(units=1,
                               activation='sigmoid' )
model = tf.keras.Sequential()
model.add(input_layer)
model.add(dense1)
model.add(dense2)
model.summary()



#print(tf.__version__)

```

```python
model.compile(optimizer=tf.keras.optimizers.SGD(0.9),loss='mse',metrics=['accuracy'])
#tf.keras.optimizers.SGD(0.9) 러닝메이트 1보다 작아야함다
```

```python
import tensorflow as tf
import numpy as np

# 1. 데이터 다시 로드 (변수가 꼬였을 수 있으니 새로 불러오는 게 안전합니다)
(train_images, train_labels), (test_images, test_labels) = tf.keras.datasets.mnist.load_data()

# 2. 이미지 데이터 전처리 (X 변수)
# 이미지를 784로 펴고 255로 나눕니다.
X_train = train_images.reshape(60000, 28*28) / 255.0
X_test = test_images.reshape(10000, 28*28) / 255.0

# 3. 정답 데이터 (y 변수) - 건드리지 않고 그대로 사용합니다.
y_train = train_labels # [5, 0, 4, ...] 형태의 숫자 데이터
y_test = test_labels

# 4. 모델 설계
model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(784,)),
    tf.keras.layers.Dense(512, activation='relu'),
    tf.keras.layers.Dense(256, activation='relu'),
    tf.keras.layers.Dense(10, activation='softmax')
])

# 5. 컴파일
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

# 6. 학습 (변수 이름 확인: X_train, y_train)
model.fit(X_train, y_train, epochs=10)
```
