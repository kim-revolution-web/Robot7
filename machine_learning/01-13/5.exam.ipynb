```python
from sklearn.datasets import make_blobs
X, Y = make_blobs(n_samples= 400, n_features=2, centers=3, cluster_std= 0.5, random_state= 42)
z, c = make_blobs(n_samples= 400, n_features=2, centers=3, cluster_std= 0.9, random_state= 42)
#데이터 샘플 400개 #특성 2개 군집 3개 표준편차 0.5 랜덤 결과가 고정
#군집 데이터를 인위적으로 만들어줌
print(X.shape, Y.shape)
print(z.shape, c.shape)
```

```python
print('X',X[:5]) #앞에 5개 군집 샘플 확인
print('z',z[:5])
```

```python
print('Y',Y[:5])
print('c',c[:5]) # 안체 5개 군집 라벨 확인
```

```python
import pandas as pd
cluster_df1 = pd.DataFrame(data = X, columns=['feature1', 'feature2'])
#x,y 기준으로 점 찍기
cluster_df1['target'] = Y  #정답 새컬럼에 붙이기
cluster_df1.head()
```

```python
cluster_df2 = pd.DataFrame(data = z, columns=['feature3', 'feature4'])
cluster_df2['target'] = c
cluster_df2.head()
```

```python
cluster_df1.plot.scatter(x='feature1', y='feature2',c='target',colormap='viridis')
cluster_df2.plot.scatter(x='feature3', y='feature4',c='target',colormap='viridis')
#c 는 색이고 target 별로 색을 다르게 함 viridis 팔레트 색을 바꿔줌
```
